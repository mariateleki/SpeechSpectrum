```python
MEDIAN_MIN = 51
P25_MIN    = 37
P75_MIN    = 63
PAY        = 13.00
FAIR_PAY   = 12.00

for label, m in [('25th', P25_MIN), ('median', MEDIAN_MIN), ('75th', P75_MIN)]:
    rate = PAY / (m / 60)
    flag = '  ⚠️ below fair-pay' if rate < FAIR_PAY else ''
    print(f'{label:>6s}: {m} min → ${rate:.2f}/hr{flag}')
```
