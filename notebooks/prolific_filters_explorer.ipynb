# Prolific filters explorer

Refresh: `python scripts/fetch_prolific_filters.py` (reads `PROLIFIC_TOKEN` from `.env`, writes `data/prolific_filters.json`).

```python
import json
from pathlib import Path
import pandas as pd
pd.set_option('display.max_colwidth', 200); pd.set_option('display.max_rows', 200)

filters = json.loads(Path('../data/prolific_filters.json').read_text())
print(f'{len(filters)} filters loaded')

def _n(c): return len(c) if isinstance(c, (dict, list)) else 0
summary = pd.DataFrame([{
    'filter_id': f.get('filter_id'),
    'title':     f.get('title'),
    'n_choices': _n(f.get('choices')),
    'question':  (f.get('question') or '').strip().replace('\n',' ')[:160],
} for f in filters])
summary
```

```python
# Drill into one filter's full choice list. Change FILTER_ID to inspect another.
FILTER_ID = 'employment-sector'

f = next((x for x in filters if x.get('filter_id') == FILTER_ID), None)
if f is None:
    print(f'No filter named {FILTER_ID!r}.')
else:
    print(f'{f["filter_id"]}  —  {f["title"]}')
    print(f'Q: {(f.get("question") or "").strip()}\n')
    choices = f.get('choices') or {}
    items = (sorted(choices.items(), key=lambda kv: int(kv[0]) if kv[0].isdigit() else kv[0])
             if isinstance(choices, dict) else list(enumerate(choices)))
    print(f'{len(items)} choices:')
    for k, v in items:
        print(f'  [{k:>3}] {v}')
```
